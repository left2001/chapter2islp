```python
rng3=np.random.default_rng(seed=3)
y=rng3.standard_normal(10)
#Mean of y ways to get it 
np.mean(y)
y.mean()
#Variance of y ways to get it 
np.var(y)
y.var()
np.mean((y-y.mean())**2)  
#Standard deviation of y ways to get it
np.std(y)
y.std()
np.sqrt(np.var(y))
```

```python
#We can apply np.mean(), np.var(), and np.std() to a matrix along a specific axis. For example, if we have a 3x4 matrix, we can compute the mean, the variance, and standard deviation along the rows (axis=0) or along the columns (axis=1).
X=rng3.standard_normal((3,4))
X
X.mean(axis=0) #mean along the rows
X.mean(axis=1) #mean along the columns
X.var(axis=0) #variance along the rows
X.var(axis=1) #variance along the columns
X.std(axis=0) #standard deviation along the rows
X.std(axis=1) #standard deviation along the columns
```

INDEXING DATA

```python
A=np.array(np.arange(16)).reshape(4,4)
A
A[1,2].item() #accessing the element in the second row and third column of the matrix A
A[[1,2]]#accessing the second and third rows of the matrix A
A[:,[1,2]]#accessing the second and third columns of the matrix A
```

```python
#Now suppose we want to access the submatrix of A that consists of the second and third rows and columns. Naturally, we woyuld like to use the following syntax:
A[[1,2],[1,2]]
#But oops!
# We get the same result as np.array([A[1,1], A[2,2]]) which is not what we want. 
A[[1,3],[0,2,3]]
#By trying the above syntax, we gert an IndexError because the two lists of indices have different lengths. 
```

```python
#To get the desirable submatrix we use the following syntax:
A[[1,2]][:,[1,2]]
#We can also use the np.ix_() function to get the same result:
A[np.ix_([1,2],[1,2])]
```

```python
A[1:4:2,1:4:2] 
```

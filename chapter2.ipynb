```python
x=np.array([[1,2],[3,4],[5,6]])
print("og x:\n",x)
#reshape() function
x_reshaped=x.reshape(2,3)
print("reshaped x:\n",x_reshaped)
```

```python
#Elements' access 
x_reshaped[0,1] #first row, second column
```

```python
print("og x:\n",x)
print("reshaped x:\n",x_reshaped)
x_reshaped[0,0] = 999 #we change the first row, first column element of the reshaped matrix
print("reshaped x after change:\n",x_reshaped)
print("og x after reshaped change:\n",x) #the original matrix is also changed because reshaping does not create a copy of the original matrix
```

The change in the element of x happens because we don't make a new copy of the matrix  
We use the same memory space to store them  
We can say that those are the ecos of python

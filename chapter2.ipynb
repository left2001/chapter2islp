```python
rng3=np.random.default_rng(seed=3)
y=rng3.standard_normal(10)
#Mean of y ways to get it 
np.mean(y)
y.mean()
#Variance of y ways to get it 
np.var(y)
y.var()
np.mean((y-y.mean())**2)  
#Standard deviation of y ways to get it
np.std(y)
y.std()
np.sqrt(np.var(y))
```

```python
#We can apply np.mean(), np.var(), and np.std() to a matrix along a specific axis. For example, if we have a 3x4 matrix, we can compute the mean, the variance, and standard deviation along the rows (axis=0) or along the columns (axis=1).
X=rng3.standard_normal((3,4))
X
X.mean(axis=0) #mean along the rows
X.mean(axis=1) #mean along the columns
X.var(axis=0) #variance along the rows
X.var(axis=1) #variance along the columns
X.std(axis=0) #standard deviation along the rows
X.std(axis=1) #standard deviation along the columns
```

INDEXING DATA

```python
A=np.array(np.arange(16)).reshape(4,4)
A
A[1,2].item() #accessing the element in the second row and third column of the matrix A
A[[1,2]]#accessing the second and third rows of the matrix A
A[:,[1,2]]#accessing the second and third columns of the matrix A
```

```python
#Now suppose we want to access the submatrix of A that consists of the second and third rows and columns. Naturally, we woyuld like to use the following syntax:
A[[1,2],[1,2]]
#But oops!
# We get the same result as np.array([A[1,1], A[2,2]]) which is not what we want. 
A[[1,3],[0,2,3]]
#By trying the above syntax, we gert an IndexError because the two lists of indices have different lengths. 
```

```python
#To get the desirable submatrix we use the following syntax:
A[[1,2]][:,[1,2]]
#We can also use the np.ix_() function to get the same result:
A[np.ix_([1,2],[1,2])]
```

```python
A[1:4:2,1:4:2] 
```

BOOLEAN INDEXING 

```python
np.shape(A)
keep_rows=np.zeros(A.shape[0],dtype=bool)
keep_rows
keep_cols=np.zeros(A.shape[1],dtype=bool)
keep_cols
keep_rows[[1,3]]=True#Set the second and fourth rows to be kept
keep_rows
notation01={"check": keep_rows==np.array([0, 1, 0, 1])}
bool(np.all(notation01["check"]))#Check if all elements of the "check" array are True
bool(np.any(notation01["check"]))#Check if any elements of the "check" array are True
```

```python
A[keep_rows]
keep_cols = np.zeros(A.shape[1], bool)
keep_cols
keep_cols[[1,3]]=True#Set the second and fourth columns to be kept
keep_cols
idx_bool= np.ix_(keep_rows, keep_cols)
idx_bool
A[idx_bool]
idx_mixed = np.ix_([1,3], keep_cols)
A[idx_mixed]
```

```python
#Check for missing values in an entire column
missing_values = (Auto["horsepower"] == "?").sum() #Count the number of missing values in the "horsepower" column of the Auto dataset. The "?" character is used to represent missing values in this dataset, so we count how many times it appears in the "horsepower" column.
print(f"Number of missing values in 'horsepower' column: {missing_values}")
#Clean the dataset by removing rows with missing values in the "horsepower" column
Auto_clean = Auto[Auto["horsepower"] != "?"] #We make a new Dataframe that only contains rows where the "horsepower" column does not have missing values AND we dont want to change the original Dataframe Auto. We use the != operator to select rows where the "horsepower" column is not equal to "?".
print(f"Number of rows after cleaning: {Auto_clean.shape[0]}")
```

```python
#Let's create a dataframe consisting of the weight and  origin of the subset of cars with year greater than 80
idx_80 = Auto_clean["year"] > 80
Auto_clean.loc[idx_80,["weight","origin"]]
```

```python
#Using the lambda function
Auto_clean.loc[lambda data: data["year"]>80,["weight","origin"]]
```

```python
#We will take the subset of cars from the upper dataset 
Auto_clean.loc[lambda data: (data["year"]>80) & (data["mpg"]>30),["weight","origin"]]
#The subset of all Ford and Datsun cars with displacement less than 300
Auto_clean.loc[lambda data: (data["displacement"]<300) & (data.index.str.contains("ford") | data.index.str.contains("datsun")), ["weight","origin"]]
```
